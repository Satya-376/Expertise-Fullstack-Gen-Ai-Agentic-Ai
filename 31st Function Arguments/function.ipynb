# keyword argument 

```python
def person(name, age): 
    print(name)
    print(age + 1)
    
person(age = 23, name = 'nit') 
```

```python
def person(name, age, mob): 
    print(name)
    print(age + 1)
    print(mob)
    
person(age = 23, name = 'nit', 12345 )  
```

# variable length

```python
def sum(a, b):
    c = a+b
    print(c)
    
sum(5,6) 
```

```python
def sum(a, b):
    c = a+b
    print(c)
    
sum(5,6,7,8,9,10) 
```

```python
def sum(a, *b):
    c = a+b
    print(c)
    
sum(5,6,7,8,9,10) 
```

```python
def sum(a, *b): # 1st argument is fixed but for 2nd argument
    #c = a+b
    print(type(a))
    print(type(b))

sum(5,6,7,8) 
```
